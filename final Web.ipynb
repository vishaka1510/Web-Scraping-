```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
 #we want (name,price,rating,discription)
Product_name=[]
Product_Prices=[]
Product_Review=[]
Product_description=[]


for i in range(2,12):
    url="https://www.flipkart.com/search?q=mobile+under+50000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page=1"+str(i)
    page=requests.get(url)
    page
    
    #we should always get this respons 200
    soup=BeautifulSoup(page.text,"lxml")
    box=soup.find("div",class_="DOjaWF gdgoEp")
    
    
    Names=box.find_all("div",class_="KzDlHZ")
    print(Names)
    
    for i in Names:
        name=i.text
        Product_name.append(name)
    print(Product_name)  
    
    print(len(Product_name))
    
    Price=box.find_all("div",class_="Nx9bqj _4b5DiR")
    print(Price)
    
    for i in Price:
        price=i.text
        Product_Prices.append(price)
    print(Product_Prices)  
    
    print(len(Product_Prices))
    
    Description=box.find_all("ul",class_="G4BRas")
    print(Description)
    
    for i in Description:
        description=i.text
        Product_description.append(description)
    print(Product_description)
    
    print(len(Product_description))
    
    reviews=box.find_all("div",class_="XQDdHH")
    print(reviews)
    
    print(len(reviews))
    
    for i in reviews:
        Reviews=i.text
        Product_Review.append(Reviews)
    print(Product_Review)   
    
df=pd.DataFrame({"Product Name":Product_name,"Product Price":Product_Prices,"Product Review":Product_Review,"Product description":Product_description})
# print(df)


df.to_csv("C:/Users/VISHAKA/OneDrive/DATA ANALYST/PYTHON/Flipkart_mobiles_under_50000_get.csv")
```
